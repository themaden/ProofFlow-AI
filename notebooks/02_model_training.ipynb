# 02 — Model Training

Preprocessing and baseline comparison with a reproducible stratified holdout.

**Data:** Synthetic invoice and purchase-order records. No production data. Not a bank credit scoring model.

```python
from pathlib import Path
import sys
root = Path.cwd().resolve()
if root.name == "notebooks": root = root.parent
sys.path.insert(0, str(root / "src"))
```

```python
from proofflow.ml import synthetic_invoices, split_data, build_pipeline
from sklearn.dummy import DummyClassifier
from sklearn.metrics import roc_auc_score, average_precision_score
df = synthetic_invoices()
x_train, x_test, y_train, y_test = split_data(df)
baseline = DummyClassifier(strategy="prior").fit(x_train, y_train)
model = build_pipeline().fit(x_train, y_train)
for title, candidate in [("baseline",baseline),("logistic",model)]:
    proba = candidate.predict_proba(x_test)[:,1]
    print(title, "ROC-AUC", round(roc_auc_score(y_test,proba),3), "AP", round(average_precision_score(y_test,proba),3))
```

**Leakage note:** The generator defines the label using information also present in the features. This is suitable for workflow testing, **not** evidence that a learned model predicts real-world fraud, credit risk or independent outcomes.
