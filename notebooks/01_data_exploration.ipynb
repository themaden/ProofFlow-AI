# 01 — Exploratory Data Analysis

Inspect distributions, class balance and synthetic feature relationships.

**Data:** Synthetic invoice and purchase-order records. No production data. Not a bank credit scoring model.

```python
from pathlib import Path
import sys
root = Path.cwd().resolve()
if root.name == "notebooks": root = root.parent
sys.path.insert(0, str(root / "src"))
```

```python
from proofflow.ml import synthetic_invoices
import matplotlib.pyplot as plt
df = synthetic_invoices()
print(df.head().to_string(index=False))
print("Missing fields:", df.isna().sum().to_dict())
print("Class balance:", df.needs_review.value_counts(normalize=True).round(3).to_dict())
```

```python
fig, axes = plt.subplots(1,2,figsize=(10,3))
df.needs_review.value_counts().sort_index().plot.bar(ax=axes[0],title="Review label counts")
df.assign(relative_amount=(df.invoice_amount / df.po_amount)).groupby("needs_review").relative_amount.plot.hist(ax=axes[1],bins=35,alpha=.55,legend=True)
axes[1].set_title("Amount ratio by label"); plt.tight_layout(); plt.show()
```

**Interpretation:** The label is generated using invoice amount mismatches and a recency/tenure rule, not manually annotated financial risk.
